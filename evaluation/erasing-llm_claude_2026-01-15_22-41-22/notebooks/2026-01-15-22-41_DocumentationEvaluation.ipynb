```python
# Now let me perform a detailed comparison to evaluate DE1-DE3 criteria

# Key metrics to compare:
# Original (from plan.md):
# - WMDP-Bio accuracy: 29.7-33.7%
# - WMDP-Cyber accuracy: 26.6-28.2%
# - MMLU: 75.2-78.8%
# - MT-Bench: 7.1-7.9
# - R-PPL: 4.3-10.9

# Replicated (from documentation_replication.md):
# - WMDP-Bio: 27.9%
# - WMDP-Cyber: 29.2%
# - Harry Potter: 66.4% (not directly comparable, but shows preservation)

# Calculate deviations
def calculate_deviation(replicated, expected_min, expected_max):
    """Calculate deviation from expected range"""
    expected_mid = (expected_min + expected_max) / 2
    deviation = abs(replicated - expected_mid) / expected_mid * 100
    return deviation, expected_mid

# WMDP-Bio
bio_replicated = 27.9
bio_expected_min, bio_expected_max = 29.7, 33.7
bio_deviation, bio_mid = calculate_deviation(bio_replicated, bio_expected_min, bio_expected_max)

# WMDP-Cyber
cyber_replicated = 29.2
cyber_expected_min, cyber_expected_max = 26.6, 28.2
cyber_deviation, cyber_mid = calculate_deviation(cyber_replicated, cyber_expected_min, cyber_expected_max)

print("=== RESULT FIDELITY ANALYSIS ===")
print(f"\nWMDP-Bio:")
print(f"  Replicated: {bio_replicated}%")
print(f"  Expected: {bio_expected_min}% - {bio_expected_max}%")
print(f"  Mid-point: {bio_mid}%")
print(f"  Deviation from mid: {bio_deviation:.2f}%")
print(f"  Status: {'WITHIN 5%' if bio_deviation <= 5 else 'EXCEEDS 5%'}")

print(f"\nWMDP-Cyber:")
print(f"  Replicated: {cyber_replicated}%")
print(f"  Expected: {cyber_expected_min}% - {cyber_expected_max}%")
print(f"  Mid-point: {cyber_mid}%")
print(f"  Deviation from mid: {cyber_deviation:.2f}%")
print(f"  Status: {'WITHIN 5%' if cyber_deviation <= 5 else 'EXCEEDS 5%'}")
```

```python
# Let me reconsider the deviation calculation
# The prompt says "within 5% deviation" - let me interpret this more carefully
# A more standard interpretation might be absolute percentage point difference
# or deviation from the expected range boundaries

def check_within_range_tolerance(replicated, expected_min, expected_max, tolerance_pct=5):
    """
    Check if replicated value is within tolerance of expected range.
    Tolerance is applied to the range boundaries.
    """
    # Calculate absolute tolerance based on range midpoint
    range_mid = (expected_min + expected_max) / 2
    tolerance = range_mid * tolerance_pct / 100
    
    # Extended range with tolerance
    lower_bound = expected_min - tolerance
    upper_bound = expected_max + tolerance
    
    within_tolerance = lower_bound <= replicated <= upper_bound
    
    return within_tolerance, lower_bound, upper_bound

# Alternative: check deviation as percentage points difference from nearest boundary
def check_percentage_point_deviation(replicated, expected_min, expected_max, max_deviation=5):
    """
    Check if replicated value is within max_deviation percentage points
    from the expected range.
    """
    if replicated < expected_min:
        deviation = expected_min - replicated
    elif replicated > expected_max:
        deviation = replicated - expected_max
    else:
        deviation = 0  # Within range
    
    within_tolerance = deviation <= max_deviation
    return within_tolerance, deviation

print("=== ALTERNATIVE DEVIATION ANALYSIS ===")
print("\n--- Method 1: 5% tolerance on range boundaries ---")

bio_tol, bio_low, bio_high = check_within_range_tolerance(bio_replicated, bio_expected_min, bio_expected_max, 5)
cyber_tol, cyber_low, cyber_high = check_within_range_tolerance(cyber_replicated, cyber_expected_min, cyber_expected_max, 5)

print(f"\nWMDP-Bio:")
print(f"  Replicated: {bio_replicated}%")
print(f"  Expected range: {bio_expected_min}% - {bio_expected_max}%")
print(f"  Extended tolerance range: {bio_low:.2f}% - {bio_high:.2f}%")
print(f"  Within tolerance: {bio_tol}")

print(f"\nWMDP-Cyber:")
print(f"  Replicated: {cyber_replicated}%")
print(f"  Expected range: {cyber_expected_min}% - {cyber_expected_max}%")
print(f"  Extended tolerance range: {cyber_low:.2f}% - {cyber_high:.2f}%")
print(f"  Within tolerance: {cyber_tol}")

print("\n--- Method 2: Percentage point deviation from nearest boundary ---")
bio_pp_tol, bio_pp_dev = check_percentage_point_deviation(bio_replicated, bio_expected_min, bio_expected_max, 5)
cyber_pp_tol, cyber_pp_dev = check_percentage_point_deviation(cyber_replicated, cyber_expected_min, cyber_expected_max, 5)

print(f"\nWMDP-Bio:")
print(f"  Deviation from nearest boundary: {bio_pp_dev:.1f} percentage points")
print(f"  Within 5 pp tolerance: {bio_pp_tol}")

print(f"\nWMDP-Cyber:")
print(f"  Deviation from nearest boundary: {cyber_pp_dev:.1f} percentage points")
print(f"  Within 5 pp tolerance: {cyber_pp_tol}")
```
